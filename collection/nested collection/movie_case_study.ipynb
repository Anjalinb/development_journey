```python
movies=[
    ['lokah','malayalam',120,2025,9],
    ['spiderman bnd','english',180,2026,10],
    ['bahubali','telungu',190,2018,9],
    ['bangalore days','malayalam',200,2014,8],
    ['blast','tamil',120,2026,7]
]
```

```python
all_movie_titles=[m[0] for m in movies]
all_movie_titles
```

```python
language=[m[1] for m in movies]
language
```

```python
movie_uppercase=[m[0].upper() for m in movies]
print(movie_uppercase)
```

```python
name_filter=[m[0] for m in movies if m[0].startswith("s")]
name_filter
```

```python
malayalam_movies=[m[0] for m in movies if m[1]=='malayalam']
malayalam_movies
```

```python
languages=[m[1] for m in movies]
lc={l:languages.count(l) for l in languages}
lc
```

```python
import matplotlib.pyplot as plt
x=lc.keys()
y=lc.values()
plt.bar(x,y)
```

```python
runtime=[m[2] for m in movies]

max_runtime_movie=[m[0] for m in movies if m[2]==max(runtime)]
max_runtime_movie
```

```python
runtime=[m[2] for m in movies]
min_runtime_movie=[m[0] for m in movies if m[2]==min(runtime)]
min_runtime_movie
```

```python

print("Average:")
sum(runtime)/len(runtime)
```
