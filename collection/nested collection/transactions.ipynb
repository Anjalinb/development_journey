MULTI COLUMN

```python
#total amount paid by each payer
payer_summary={}
for t in transactions:
    payer=t['payer']
    amount=t['amount']
    if payer in payer_summary:
        payer_summary[payer]+=amount
    else:
        payer_summary[payer]=amount
payer_summary
```

```python
#payer who paid the most
payer_sum_values=[v for v in payer_summary.values()]
highest_payer=[k for k,v in payer_summary.items() if v==max(payer_sum_values)]
highest_payer
```

```python
#total amount received by each beneficiary
beneficiary_summary={}
for t in transactions:
    beneficiary=t['beneficiary']
    amount=t['amount']
    if beneficiary in beneficiary_summary:
        beneficiary_summary[beneficiary]+=amount
    else:
        beneficiary_summary[beneficiary]=amount

beneficiary_summary
```

```python
#beneficiary who received the most
beneficiary_summary_values=[v for v in beneficiary_summary.values()]
ben_received_most=[k for k,v in beneficiary_summary.items() if v==max(beneficiary_summary_values)]
ben_received_most
```
