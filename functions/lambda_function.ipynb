```python
"""
lambda function= anonymous function with single expression
syntax: lambda p1,p2:expression
Eg:

"""
```

```python
add_num=lambda n1,n2:n1+n2
add_num(10,20)
```

```python
sub_num=lambda n1,n2:n1-n2
sub_num(30,10)
```

```python
max_two=lambda n1,n2: n1 if n1>n2 else n2
print(max_two(12,98))
```

```python
min_two=lambda n1,n2: n1 if n1<n2 else n2
min_two(43,34)
```

```python
user_exp=[
["hari",6],
["abin",1],
["bijoy",4],
["cibin",2],
["akash",7],
]
max(user_exp)
```

```python

max(user_exp,key=lambda lst:lst[1])

```

```python
min(user_exp,key=lambda lst:lst[1])
```

```python

weather=[
    {"district":"tsr","temp":38},
    {"district":"tvm","temp":35},
    {"district":"ekm","temp":42},
    {"district":"pkd","temp":49},
    {"district":"kty","temp":30}
]

max(weather,key=lambda dictionary:dictionary.get("temp"))
```

```python
min(weather,key=lambda dictionary:dictionary.get("temp"))
```
