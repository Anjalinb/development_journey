```python
movies_data = [ ["dangal", "hindi", 161, 2016, 8.3],       # 0 
                ["premam", "malayalam", 156, 2015, 8.3],   # 1
                ["inception", "english", 148, 2010, 8.8],  # 2 
                ["vikram", "tamil", 175, 2022, 8.3],       # 3 
                ["parasite", "korean", 132, 2019, 8.5],  # 4
                ["rrr", "telugu", 187, 2022, 7.8],          # 5
                ["drishyam", "malayalam", 160, 2013, 8.3], # 6
                ["avatar", "english", 162, 2009, 7.9],     # 7
                ["kumbalangi nights", "malayalam", 135, 2019, 8.5], # 8
                ["sholay", "hindi", 204, 1975, 8.1]  ]      # 9 
```

```python
titles=[m[0] for m in movies_data]
titles
```

```python
filter_year=[m[0] for m in movies_data if m[3]>2015]
filter_year
```

```python

filter_rating=[m for m in movies_data if m[4]>=8.5]
len(filter_rating)
```

```python
runtime=[m[2] for m in movies_data]
avg=sum(runtime)/len(runtime)
avg
```

```python
year_and_runtime=[m[0] for m in movies_data if m[2]>150 and m[-1]>8]
year_and_runtime
```

```python
rating=[m[-1] for m in movies_data]
highest_rating=[m for m in movies_data if m[-1]==max(rating)]
highest_rating
```

```python
year=[m[-2] for m in movies_data]
oldest_movie=[(m[0],m[-2]) for m in movies_data if m[-2]==min(year)] 
oldest_movie
```
