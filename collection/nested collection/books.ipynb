```python
books_data = [ ["The Hobbit", "Fantasy", 15, 1937, 4.8],    
               ["Dune", "Sci-Fi", 18, 1965, 4.7],
               ["Project Hail Mary", "Sci-Fi", 22, 2021, 4.9], 
               ["Atomic Habits", "Self-Help", 16, 2018, 4.8],     
                ["1984", "Dystopian", 12, 1949, 4.6],
               ["The Silent Patient", "Mystery", 14, 2019, 4.5],  # 5 
                ["Sapiens", "History", 20, 2014, 4.7],
                 ["Harry Potter", "Fantasy", 25, 1997, 4.9],   
                 ["Educated", "Memoir", 17, 2018, 4.4],            
                 ["Klara and the Sun", "Sci-Fi", 13, 2021, 4.2] ]
```

```python
titles=[b[0] for b in books_data if len(b[0])>10]
titles
```

```python
title=[b[0] for b in books_data if b[0].startswith("The")]
title
```

```python
genres={b[1] for b in books_data}
genres
```

```python
sci_fi=[b[0] for b in books_data if b[1]=="Sci-Fi"]
len(sci_fi)
```

```python
price=[(b[2]) for b in books_data]
total_price=sum(price)
total_price
```

```python
price=[(b[2]) for b in books_data]

cheapest=[(b[0],b[2]) for b in books_data if b[2]==min(price)]
cheapest
```

```python
title_and_age=[(b[0],2026-b[-2]) for b in books_data]
title_and_age
```

```python
books=[b[0] for b in books_data if b[-2]<2000]
books
```

```python
rating=[b[-1] for b in books_data]
avg=sum(rating)/len(rating)
avg
```

```python
rating_gt=[b[0] for b in books_data if b[-1]>=4.8]
len(rating_gt)
```

```python
new=[b for b in books_data if b[1]=='Sci-Fi' and b[-2]>2010]
new_prices=[b[2] for b in new]
most_expensive=[b[0] for b in new if b[2]==max(new_prices)]
most_expensive
```
